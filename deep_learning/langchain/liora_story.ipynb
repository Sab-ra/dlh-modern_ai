```python
from langchain_huggingface import HuggingFacePipeline

llm = HuggingFacePipeline.from_model_id(
    model_id='google/flan-t5-base',
    task='text2text-generation',
    pipeline_kwargs={
        "max_new_tokens": 50,
    }
)
print(llm)
```

Super, step-by-step works!!! Liora said a word. So, let's make a function like for the curriculum, and test like in the curriculum

```python
def initialize_hf_llm(
        model_name, max_tokens
):
    """
    Set up a HggF text-to-text model in LangChain
    """
    llm = HuggingFacePipeline.from_model_id(
        model_id=model_name,
        task='text2text-generation',
        pipeline_kwargs={"max_new_tokens": max_tokens}
    )
    return llm
```

```python
llm = initialize_hf_llm(
    model_name='google/flan-t5-base',
    max_tokens=50
)
print(llm.invoke(input=prompt))
```
